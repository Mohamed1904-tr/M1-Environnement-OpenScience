```python
import pandas as pd

data = pd.read_csv("/donnees_meteo.csv")

data.head()
```

## Activité 8 — Température moyenne

```python
data["Temperature"].mean()
```

**Réponse :** La température moyenne enregistrée pendant la période étudiée est de **32.64 °C**.

## Activité 9 — Température maximale et minimale

```python
data["Temperature"].max()
```

```python
data["Temperature"].min()
```

```python
print("Date du max :", data.loc[data["Temperature"].idxmax(), "Date"])
print("Date du min :", data.loc[data["Temperature"].idxmin(), "Date"])
```

**Réponses :**
- Température maximale : 39.2 °C, enregistrée le 2025-07-20
- Température minimale : 25.7 °C, enregistrée le 2025-06-12

## Activité 10 — Précipitations

```python
data["Precipitation"].mean()
```

```python
data["Precipitation"].sum()
```

**Réponse :** La quantité totale de précipitations est de **205.2 mm** sur la période étudiée (moyenne : 2.28 mm par jour).

## Activité 11 — Visualisation de la température

```python
import matplotlib.pyplot as plt

plt.plot(data["Date"], data["Temperature"])

plt.xlabel("Date")
plt.ylabel("Température (°C)")
plt.title("Évolution de la température")
plt.xticks(data["Date"][::10], rotation=45)

plt.show()
```

**Réponses :**
- La température augmente globalement de début juin jusqu'à la fin juillet, puis elle diminue en août.
- Oui, il existe des variations importantes d'un jour à l'autre (de 25.7 °C à 39.2 °C sur la période).
- Les températures les plus élevées sont observées en juillet (maximum : 39.2 °C le 2025-07-20).

## Activité 12 — Précipitations

```python
# @title Titre par défaut
plt.bar(data["Date"], data["Precipitation"])

plt.xlabel("Date")
plt.ylabel("Précipitation (mm)")
plt.title("Précipitations enregistrées")
plt.xticks(data["Date"][::10], rotation=45)

plt.show()
```

**Interprétation :**
- **Jours les plus pluvieux :** 2025-06-15 (28.1 mm), 2025-07-25 (24.7 mm), 2025-06-27 (18.8 mm) et 2025-08-17 (17.4 mm).
- **Périodes sans précipitation :** 71 jours sur 90 sans pluie. Les plus longues : du 2025-07-26 au 2025-08-07 (13 jours) et du 2025-07-03 au 2025-07-13 (11 jours).
- **Variations observées :** les pluies sont rares et irrégulières, avec quelques épisodes intenses isolés. Juillet est le mois le plus sec (50.8 mm) ; juin (75.8 mm) et août (78.6 mm) sont plus arrosés.
